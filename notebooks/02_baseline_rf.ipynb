```python
import sys
from pathlib import Path

# Add project root to path
project_root = Path.cwd().parent
sys.path.insert(0, str(project_root))

from src.data import load_data
from sklearn.model_selection import GroupKFold, cross_validate, KFold
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import root_mean_squared_error, r2_score 
```

```python
df = load_data()
```

```python
x = df.drop(['permeability', 'resistivity'], axis=1) # Independent features
y = df[['permeability', 'resistivity']]              # Dependent features
groups = df['permeability']     # Group by permeability
```

```python
# Initialize GroupkFold with 11 splits
gfk = GroupKFold(n_splits=11)

# Initialize RandomForestRegressor
rf = RandomForestRegressor()
```

```python
score={'r2':'r2', 'rmse':'neg_root_mean_squared_error'}
for target in ['permeability', 'resistivity']:
    y_single = df[target]
    scores = cross_validate(estimator=rf, X=x, y=y_single, cv=gfk, groups=groups, scoring=score)
    print(f'Average r2 Score: {scores["test_r2"].mean()}')
    print(f'RMSE Score: {scores["test_rmse"].mean()}')
```
